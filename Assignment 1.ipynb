```python
#Create a List
students_list=["Tina", "Riya", "Aryan","Sumit","Priyanka"]
print("\nStudents List:")
print(students_list)
```

```python
#Create a Tuple
students_tuple=("Tina", "Riya", "Aryan","Sumit","Priyanka")
print("\nStudents Tuple:")
print(students_tuple)
```

```python
#Create a Dictionary
student_dictionary={
    100:"Tina",
    101:"Riya",
    102:"Aryan",
    103:"Sumit",
    104:"Priyanka"
}
print("\nStudents Dictionary:")
print(student_dictionary)
```

```python
#Add a student's name to dictionary
student_dictionary[105]="Krisha"
print("\nAfter adding:")
print(student_dictionary)
```

```python
#Update a student's name
student_dictionary[100]="Mohit"

print("\nAfter updating")
print(student_dictionary)
```

```python
#Delete a student's name
del student_dictionary[102]

print("\nAfter deleting:")
print(student_dictionary)
```
