```python
#Program to check if the triangle is right angled triangle or not using functions(predefined)
def right_angled_triangle(a,b,c):
    if a*a+ b*b==c*c:
        print("Right angled triangle")
    else:
        print("Not a right angled triangle")
right_angled_triangle(3,4,5)
```

```python
#Program to check if the triangle is right angled triangle or not using functions(predefined)
def right_triangle(p,q,r):
    if p*p+ q*q==r*r:
        print("Right angled triangle")
    else:
        print("Not a right angled triangle")
side_1=int(input("Enter first side"))
side_2=int(input("Enter second side"))
side_3=int(input("Enter third side"))
right_triangle(side_1,side_2,side_3)
```
