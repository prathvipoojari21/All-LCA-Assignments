```python
#Step 4:Display passenger names
print("\nPassenger Names:")
print(df["Name"].head())

#Step 5:Count passengers by class
print("\nPassengers by Gender:")
print(df["Pclass"].value_counts())

#Step 6: Calculate average age
print("\nPassengers by Gender:")
print(df["Age"].mean())

#Step 7: Find the highest fare
print("\Highest Fare:")
print(df["Fare"].max())

#Step 8: Find the average fare
print("\nAverage Fare:")
print(df["Fare"].mean())

df=df.drop(['Cabin','Embarked'],axis=1)
```

```python
#Step 9: Filter passengers older thsn 60
print("\nPassengers Older Than 60:")
print(df[df["Age"]>60])
```

```python
#Step 10: Display passengers who survived
print("\nPassengers Who Survived:")
print(df[df["Survived"]==1][["Name","Sex","Age"]].head())
```

```python
#Step 11: Count survivors and non-survivors
print("\nSurvival Counts:")
print(df["Survived"].value_counts())
```

```python
#Step 12: Calculate survival rate by gender
print("\nSurvival Rate by Gender:")
print(df.groupby("Sex")["Survived"].mean())
```

```python
#Step 13: Sort passengers by fare
print("\nTop 10 Highest Fares:")
print(
    df.sort_values("Fare",ascending=False)
    [["Name","Fare"]]
    .head(10)
)
```

```python
#Step 14: Count passengers by passenger class
print("\nPassengers by Class:")
print(df["Pclass"].value_counts().sort_index())
```

```python
#Step 15:Calculate average age by gender
print("\nAverage Age by Gender:")
print(df.groupby("Sex")["Age"].mean())
```

```python
#Step 16: Display embarkation port counts
print("\nPassengers by Embarkation Port:")
print(df["Pclass"].value_counts())
```

```python
#Step 17: Save a copy of the dataset
df.to_csv("titanic_copy.csv",index=False)
```

```python
#Step 18: Remove duplicates
df=df.drop_duplicates()
print("Duplicates Rows Removed")
```

```python
#Step 19: Handle Missing values
for col in df.columns:
    if df[col].isnull().any():
        if pd.api.types.is_numeric_dtype(df[col]):
            df[col]=df[col].fillna(df[col].mean())
        else:
            df[col]=df[col].fillna(df[col].mode()[0])
print("\nMissing Values Handled")
```
